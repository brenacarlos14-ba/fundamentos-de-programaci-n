## EXTRA 2 CASTIGO

```python
candidatos = {
    "A":0, 
    "B":0, 
    "C":0
    }

cantidad = int(input("¿cuantos votos van a ser?"))
conteo = 1

while conteo <= cantidad:
    voto = input(f"ingresa el voto {conteo} (A, B o C): ")
    if voto.upper() in candidatos:
        print(f"voto es valido para {voto.upper()}")
        candidatos[voto.upper()]+= 1
        conteo += 1
    else:
        print("el voto no es valido, otra vez")

print("\nResultados: ")
for k, v in candidatos.items():
    print(f"opción {k}, votos:{v} ")

print(f"\nNumero de votos: {cantidad}")
```
