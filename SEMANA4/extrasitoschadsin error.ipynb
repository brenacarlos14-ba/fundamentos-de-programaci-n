# EXTRA 1

```python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

def imprimir_matriz(matriz):
    for fila in matriz:
        for elemento in fila:
            print(elemento, end=" ")
        print()

def suma_matriz(matriz):
    total = 0
    for fila in matriz:
        for numero in fila:
            total += numero
    return total

imprimir_matriz(matriz)
print(f"el resultado es: {suma_matriz(matriz)}")
```

# EXTRA 2

```python
matriz = [ 
    [3, 1, 4], 
    [4, 5, 6], 
    [2, 6, 5] ] 
suma_filas = [] 

def mostrar_tabla(matriz):
    for fila in matriz:
        for elemento in fila:
            print(elemento, end=" ")
        print()

for fila in matriz: 
    suma_actual = 0 
    for numero in fila: 
        suma_actual += numero 
    suma_filas.append(suma_actual) 

mostrar_tabla(matriz)

print(f"Suma de la fila 1: {suma_filas[0]}") 
print(f"Suma de la fila 2: {suma_filas[1]}") 
print(f"Suma de la fila 3: {suma_filas[2]}")
```

# EXTRA 3

```python
matriz = [
    [1, 2, 3], 
    [4, 5, 6], 
    [7, 8, 9], 
    [2, 4, 6]
    ]

def multiplicar_columna(matriz, columna):
    producto = matriz[0][columna]
    
    for fila in range(1, len(matriz)):
        siguiente_numero = matriz[fila][columna]
        suma_acumulada = 0
        for _ in range(siguiente_numero):
            suma_acumulada += producto
            
        producto = suma_acumulada
        
    return producto

columna = int(input("Ingresa un número (0, 1 o 2): "))

if 0 <= columna <= 2:
    resultado = multiplicar_columna(matriz, columna)
    print(f"El producto de la columna {columna} es: {resultado}")
else:
    print("Número de columna fuera de rango. Debe ser 0, 1 o 2.")
```

# EXTRA 4

```python
def generar_tabla(tamaño):
    tabla = []
    for i in range(1, tamaño + 1):
        renglón = []
        for j in range(1, tamaño + 1):
            producto = sum(i for _ in range(j)) 
            renglón.append(producto)
        tabla.append(renglón)
    return tabla

def imprimir_tabla(tabla):
    for fila in tabla:
        print(*fila, sep="\t")

tamaño = int(input("tamaño de la tabla (2 a 5): "))

if 2 <= tamaño <= 5:
    matriz = generar_tabla(tamaño)
    imprimir_tabla(matriz)
    
    r = int(input(f"renglón (1-{tamaño}): ")) - 1
    c = int(input(f"columna (1-{tamaño}): ")) - 1
    print("producto:", matriz[r][c])
else:
    print("error no chad: un tamaño entre 2 y 5.")
```
